```python
import multiprocessing
import scanpy as sc
import squidpy as sq

def run_ligrec(adata):
    # all your parameters here
    return sq.gr.ligrec(
        adata,
        n_perms=200,
        cluster_key="assigned_celltype_L0",
        copy=True,
        use_raw=False,
        transmitter_params={"categories": "ligand"},
        receiver_params={"categories": "receptor"}
    )

if __name__ == "__main__":
        multiprocessing.freeze_support()

        #adata = sq.datasets.seqfish()
        #adata = sc.read_h5ad("/Volumes/ProstateCancerEvoMain/dbs/Ongoing/Collection/MajorAnnGroups/Epithelia_SubTypes/Region1.Epithelia_SubTypes.AnnDC.h5ad")
        adata = sc.read_h5ad("/Volumes/ProstateCancerEvoMain/up-to-date/ProstateCancerEvo_SpatialAnalysis/scripts/spatial_aware_analysis/spatial_interactions/R1_MiddleLoc.h5ad")


        res = run_ligrec(adata)
res
```

```python
df_plot = res["means"]['CD8 T Cell']["tumor_cell_markers"].rename("value").reset_index()
df_plot
```

```python
import pandas as pd
import plotly.express as px

# 3. Bubble‐plot with Plotly
fig = px.scatter(
    df_plot,
    x="target",
    y="source",
    size="value",
    color="value",
    color_continuous_scale="Blues",
    size_max=20,
    hover_data=["value"],
    labels={"value": "Score"},
    width=1600,
    height=1000
)
fig.update_layout(
    xaxis_tickangle=45,
    yaxis=dict(tickfont=dict(size=8)),
    plot_bgcolor="white",
    margin=dict(l=200, r=20, t=50, b=200)
)
fig.show()
```
