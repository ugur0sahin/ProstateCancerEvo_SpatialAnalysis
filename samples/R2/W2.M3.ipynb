```r
library(monocle3)
library(ggplot2)
library(dplyr)


cds <- load_mm_data(mat_path = "/Volumes/ProstateCancerEvoMain/raw/xenium/output-XETG00283__0037793__Region_2__20241204__160314/cell_feature_matrix/matrix.mtx.gz", 
                    feature_anno_path = "/Volumes/ProstateCancerEvoMain/raw/xenium/output-XETG00283__0037793__Region_2__20241204__160314/cell_feature_matrix/features.tsv.gz", 
                    cell_anno_path = "/Volumes/ProstateCancerEvoMain/raw/xenium/output-XETG00283__0037793__Region_2__20241204__160314/cell_feature_matrix/barcodes.tsv.gz")



cell_guide <- read.csv(
  "/Volumes/ProstateCancerEvoMain/dbs/Completed/AllRegions/CellTypes/Region1/Region1.tumor_cell_markers.raw.annotated.V2.guide.csv",
  stringsAsFactors = FALSE
)

#cell_guide

keep_cells <- cell_guide$cell_id
#keep_cells


library(SummarizedExperiment)  # for colData()
meta <- as.data.frame(colData(cds))

cells_in_cds <- rownames(meta)
cells_to_keep <- intersect(cells_in_cds, keep_cells)

cds_filtered <- cds[, cells_to_keep]

#length(colnames(cds_filtered))                       # # cells now
#all(colnames(cds_filtered) %in% keep_cells)          # should be TRUE
```

```r
rownames(cell_guide) <- cell_guide$cell_id
guide_sub <- cell_guide[colnames(cds_filtered), , drop = FALSE]
colData(cds_filtered)$leiden <- guide_sub$leiden_L0


cds_filtered <- preprocess_cds(cds_filtered, num_dim = 50)
cds_filtered <- reduce_dimension(cds_filtered)

colData(cds_filtered)$leiden <- factor(
  colData(cds_filtered)$leiden,
  levels = sort(unique(guide_sub$leiden))
)
```

```r
cds_filtered <- cluster_cells(cds_filtered)
cds_filtered <- learn_graph(cds_filtered)

#plot_cells(cds_filtered, color_cells_by = "partition")
plot_cells(cds_filtered, label_groups_by_cluster=FALSE,  color_cells_by = "leiden")
```

```r
cds_sub <- choose_graph_segments(cds_filtered)

cds_3d <- reduce_dimension(cds_filtered, max_components = 3)
cds_3d <- cluster_cells(cds_3d)
cds_3d <- learn_graph(cds_3d)

cds_3d_plot_obj <- plot_cells_3d(cds_3d, color_cells_by="leiden")
cds_3d_plot_obj
```
