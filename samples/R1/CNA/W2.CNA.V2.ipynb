```r
library(infercnv)

raw_counts_matrix <- "/Users/ugursahin/ProstateCancerEvo_SpatialAnalysis/samples/R1/CNA/R1.pseudobulk.healthy_tumor.tsv"
counts <- read.delim(raw_counts_matrix, header=TRUE, row.names=1, check.names=FALSE)
counts


annotations_f <- "/Users/ugursahin/ProstateCancerEvo_SpatialAnalysis/samples/R1/CNA/R1.ann.V2.tsv"
annotations <- read.delim(annotations_f, header=TRUE, row.names=1, check.names=FALSE)
annotations

gene_order_f <- "/Users/ugursahin/ProstateCancerEvo_SpatialAnalysis/samples/R1/CNA/Xenium.Sapiens.Ref.Ordered.V2.tsv"
gene_order <- read.delim(gene_order_f, header=TRUE, row.names=1, check.names=FALSE)
gene_order


infercnv_obj = CreateInfercnvObject(raw_counts_matrix=counts,
                                    annotations_file=annotations,
                                    delim="\t",
                                    gene_order_file=gene_order,
                                    ref_group_names=c("basal_cell_of_prostate_epithelium","basal_epithelial_cell_of_prostatic_duct")) 

```

```r
infercnv_obj = infercnv::run(infercnv_obj,
                             cutoff=1, # cutoff=1 works well for Smart-seq2, and cutoff=0.1 works well for 10x Genomics
                             out_dir="/Users/ugursahin/ProstateCancerEvo_SpatialAnalysis/samples/R1/CNA", 
                             cluster_by_groups=TRUE, 
                             denoise=TRUE,
                             HMM=TRUE)
```
